## Part 1: The BankAccount class

Your task is to write in the file `work.py` the definition of two classes:

- `BankAccount` that represents some features of a real checking (or current) account
- `MinimumBalanceAccount` as a subclass of `BankAccount`

A `BankAccount` object has three attributes

- `iban` (string) (International Bank Account Number, unique)
- `name` (string) (name of the account holder)
- `balance` (float, default = 0)

The `BankAccount` class implements the following methods

- `get_balance`, which returns the balance of the account
- `get_name`, which returns the name of the account holder
- `get_iban`, which returns the iban of the account
- `deposit`, which takes a positive parameter `amount` (float). `deposit` increases the balance by `amount` and returns `amount`. If `amount` is not positive, `deposit` raises a `ValueError`
- `withdraw`, which takes a positive parameter `amount` (float). If the balance is greater than or equal to amount, `withdraw` reduces the balance by the amount specified and returns `amount`. If not, it raises a `ValueError`. If `amount` is not positive, `deposit` raises a `ValueError`.
- `transfer`, which takes two parameters:
  - `destination_account` (BankAccount)
  - `amount` (float, positive).
  `transfer` simulates the transfer of money from the instance to the `destination_account`. `transfer` withdraws `amount` from the instance, deposits `amount` in the `destination_account`, and returns `amount`. If the balance of the instance is less than `amount`, `transfer` raises a `ValueError`

## Part 2: The MinimumBalanceAccount class

The `MinimumBalanceAccount` is a subclass of `BankAccount`. It accepts two additional attributes:
- `minimum_balance` (float, default= 100, positive)
- `initial_deposit` (float, default = 100). 

The purpose of this class is to represent accounts which balance cannot go under a minimum amount (usually, these accounts are called *Savings Accounts*). The constructor should raise a `ValueError` if `initial_deposit` is less than `minimum_balance` (otherwise, the account would be created in a state of error) or if `minimum_balance` is not positive.

The `MinimumBalanceAccount` implements two additional methods:
- `set_minimum_balance`, which takes a parameter amount (float, positive). The methods sets the minimum balance to the specified amount only if amount is less than or equal to the balance of the instance (otherwise, the balance would be under the minimum). In this case, it returns `amount`. If the method cannot set the minimum balance or if `amount` is not positive, a ValueError is raised.
- `get_minimum_balance`, which returns the minimum balance

The `MinimumBalanceAccount` also overwrites `withdraw` in the base class. The `withdraw` method in `MinimumBalanceAccount` behaves like the original withdraw method and checks whether a withdrawal attempt would set the balance below the minimum or not. If this is the case, it raises a ValueError.

```python
# TEST SECTION
# DO NOT MODIFY THIS CELL

def must_raise(exception_type, f, *args):
    got_exc = None
    try:
        f(*args)
    except exception_type as exc:
        got_exc = exc
    except Exception as exc:
        raise Exception(f"was expecting a {exception_type.__name__}, got a {type(exc).__name__} instead") from exc
    if got_exc is None:
        raise Exception(f"No exception raised, was expecting a {exception_type.__name__}")
    return got_exc
```
